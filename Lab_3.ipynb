<b>Explanation</b><br>
The Fibonacci sequence is defined as:<br>

F(0) = 0, F(1) = 1<br>
F(n) = F(n-1) + F(n-2)<br>


### a) Recursive (Inefficient)

```python
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n-1) + fib_recursive(n-2)
```

```python
# Example usage:
n=10
print(f"The {n}th fibonacy number is = {fib_tab(n)}")  
```

### b) Memoization (Top-Down DP)

```python
def fib_memo(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        memo[n] = n
    else:
        memo[n] = fib_memo(n-1, memo) + fib_memo(n-2, memo)
    return memo[n]
```

```python
# Example usage:
n=10
print(f"The {n}th fibonacy number is = {fib_tab(n)}")  
```

### c) Tabulation (Bottom-Up DP)

```python
def fib_tab(n):
    dp = [0, 1] + [0]*(n-1) 
    for i in range(2, n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
```

```python
# Example usage:
n=10
print(f"The {n}th fibonacy number is = {fib_tab(n)}")  
```

### 2. 0/1 Knapsack Problem
Description:<br>
Given weights and values, determine the maximum value achievable in a knapsack of capacity W.<br>

### 1. Recursive Solution (No Memoization)

```python
def knapsack_recursive(weights, values, capacity, n):
    # Base case
    if n == 0 or capacity == 0:
        return 0
    # If the weight of the nth item is more than capacity, skip it
    if weights[n-1] > capacity:
        return knapsack_recursive(weights, values, capacity, n-1)
    else:
        # include or exclude the nth item
        included = values[n-1] + knapsack_recursive(weights, values, capacity-weights[n-1], n-1)
        excluded = knapsack_recursive(weights, values, capacity, n-1)
        return max(included, excluded)
```

```python
# Example usage:
weights = [2, 3, 4, 5]
values = [3, 4, 5, 6]
capacity = 5
n = len(values)
print(knapsack_recursive(weights, values, capacity, n))  
```

### 2. Top-Down DP with Memoization

```python
def knapsack_memo(weights, values, capacity, n, memo=None):
    if memo is None:
        memo = {}
    if (n, capacity) in memo:
        return memo[(n, capacity)]
    if n == 0 or capacity == 0:
        result = 0
    elif weights[n-1] > capacity:
        result = knapsack_memo(weights, values, capacity, n-1, memo)
    else:
        included = values[n-1] + knapsack_memo(weights, values, capacity-weights[n-1], n-1, memo)
        excluded = knapsack_memo(weights, values, capacity, n-1, memo)
        result = max(included, excluded)
    memo[(n, capacity)] = result
    return result
```

```python
# Example usage:
print(knapsack_memo(weights, values, capacity, n)) 
```

### 3. Bottom-Up DP with Table Reconstruction
Assuming 'n' is the number of items and 'capacity' is the maximum weight capacity

dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

#### Explanation:
#### This creates a list of lists (a 2D grid/matrix) 
#### with dimensions (n+1) rows and (capacity+1) columns.
#### Every cell in the grid is initialized to the value 0.

```python
def knapsack_bottom_up(weights, values, capacity):
    n = len(values)
    dp = [[0 for _ in range(capacity+1)] for _ in range(n+1)]

    # Build table dp[][] in bottom-up manner
    for i in range(1, n+1):
        for w in range(capacity+1):
            if weights[i-1] > w:
                dp[i][w] = dp[i-1][w]
            else:
                dp[i][w] = max(dp[i-1][w], values[i-1] + dp[i-1][w-weights[i-1]])

    # Reconstruction of selected items
    selected_items = []
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i-1][w]:
            selected_items.append(i-1)  # Store the index of the item
            w -= weights[i-1]

    selected_items.reverse()  # Optional: in order of input
    return dp[n][capacity], selected_items
```

```python
# Example usage:
max_value, items = knapsack_bottom_up(weights, values, capacity)
print("Max value:", max_value)             # Output: 7
print("Selected item indices:", items)     # Output: [0, 1]
```

```python
# Example usage:
weights = [2, 3, 4, 5]
values = [3, 4, 5, 6]
capacity = 5
print(knapsack(weights, values, capacity))  # Output: (7, [0, 1])
```

### 3. Longest Common Subsequence (LCS)
Description:<br>
* Find the length (and sequence) of the longest subsequence present in both strings.<br>

<b>Lab Tasks:</b><br>

* Implement DP for LCS length.<br>
* Reconstruct and print the actual LCS.<br>
* Test with different string pairs.<br>

```python
def lcs(X, Y):
    m, n = len(X), len(Y)
    dp = [[0]*(n+1) for _ in range(m+1)]
    # Fill DP table
    for i in range(m):
        for j in range(n):
            if X[i] == Y[j]:
                dp[i+1][j+1] = dp[i][j] + 1
            else:
                dp[i+1][j+1] = max(dp[i][j+1], dp[i+1][j])
    # Reconstruct LCS
    lcs_str = []
    i, j = m, n
    while i > 0 and j > 0:
        if X[i-1] == Y[j-1]:
            lcs_str.append(X[i-1])
            i -= 1
            j -= 1
        elif dp[i-1][j] > dp[i][j-1]:
            i -= 1
        else:
            j -= 1
    return dp[m][n], ''.join(reversed(lcs_str))
```

```python
# Example
X = "AGGTAB"
Y = "GXTXAYB"
print(lcs(X, Y))  
```

### 4. Coin Change (Minimum Coins)
Description:<br>
Given coin denominations and a target amount, find the minimum number of coins needed to make up that amount.<br>

<b>Lab Tasks:</b><br>

* Implement DP solution for minimum coins.<br>
* Output the coins used.<br>


```python
def coin_change(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    for i in range(1, amount+1):
        for c in coins:
            if i - c >= 0:
                dp[i] = min(dp[i], dp[i-c] + 1)
    return dp[amount] if dp[amount] != float('inf') else -1
```

```python
# Example
coins = [1, 3, 4]
amount = 6
print(coin_change(coins, amount))  # Output: 2  (using coins 3 and 3 or 4 and 2)
```
